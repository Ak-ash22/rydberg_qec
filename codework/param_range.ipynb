```julia
using SymPy, LinearAlgebra, Plots
```

```julia
Ω, δ, N, α, β, V_nn = symbols("Ω δ N α β V_nn", real=true, positive=true)
Δ0 = symbols("Δ0", real=true, positive=true, greater_than = V_nn);
```

```julia
ħ = 1.0545718e-34
β = Δ0/V_nn
α = δ / (Ω^2)

P_rabi = (Ω^2) / (Ω^2 + (Δ0 - V_nn)^2)
P_lz = exp(- Ω^2 / δ)

threshold_rabi = Eq(P_rabi,1e-3)
threshold_lz = Eq(P_lz,1e-4);

#Chosen Paramters
Omega_val = 1
N_val = 2
V_nn_val = 1000

#Solve for the values of alpha and Δ0
delta_soln = solve(subs(threshold_lz,Ω=>Omega_val,N=>N_val,V_nn=>V_nn_val),δ)[1]
delta0_soln = solve(subs(threshold_rabi,Ω=>Omega_val,N=>N_val,V_nn=>V_nn_val),Δ0)

δ_val = delta_soln
Δ0_val = [sol for sol in delta0_soln if sol > V_nn_val][1]

#Solve for the Parameters
beta_soln = Δ0_val / V_nn_val
alpha_soln = δ_val / (N_val * Omega_val^2)

#Solve for sweep time
T_sweep = 2 * (Δ0_val - V_nn_val) / δ_val

#Solve for threshold decay rate
γ_threshold = 1 / T_sweep

println("For the chosen parameters: Ω = $Omega_val, N = $N_val, V_nn = $V_nn_val :
    α = $alpha_soln
    β = $beta_soln
    Δ0 = $Δ0_val
    δ = $δ_val
    T_sweep = $T_sweep
    γ_threshold = $γ_threshold")
```

##### Notice that the Numerical values obtained follows all the constraints, including the the constraint that P_rabi = 1e-3 and P_lz = 1e-4
