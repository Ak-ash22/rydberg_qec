```julia
using LinearAlgebra, DifferentialEquations, SpecialFunctions, Plots, QuadGK
```

```julia
# Define the Whittaker W function using the integral representation
function whittakerW_integral(α,γ,V=1.0)
    # Exponential prefactor
    κ = 1im * V^2/α
    μ = -1/2
    z = -1im*γ^2/α

    prefactor = exp(-z / 2) * z^κ

    # Define the integrand
    function integrand(t)
        t^(κ - μ - 1) * (1 + t)^(κ + μ - 1) * exp(-z * t)
    end

    # Perform the integration over [0, ∞)
    integral, _ = quadgk(integrand, 0, Inf; atol=1e-8, rtol=1e-8)

    # Combine prefactor and integral
    return prefactor * integral
end


function compute_I(α,γ,V=1.0)
    #compute the integral
    w = whittakerW_integral(V,α,γ)

    #compute the prefactor
    I = (2 * pi * V^2 * exp(-pi*V^2/α) * abs(w)^2) / α

    return I
end
```

```julia
function I_analytic(α,γ,V=1.0)
    return (2 * pi * γ * V^2)/α * (1 - exp(-2* pi* V^2/α))/(γ^2 + 4* V^2)^0.5
end
```
