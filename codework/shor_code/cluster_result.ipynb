```julia
function Ht(t)
    if t<T1 || t==T1
        coeffs = coeff1[1](t)
        for i in eachindex(coeffs)
            H1.factors[i] = coeffs[i]
        end
        # return H1

    elseif t<(T1+T2) || t==(T1+T2)
        coeffs = coeff2[1](t-T1)
        for i in eachindex(coeffs)
            H2.factors[i] = coeffs[i]
        end
        # return H2

    elseif t<(T1+T2+T3) || t==(T1+T2+T3)
        coeffs = coeff3[1](t-T1-T2)
        for i in eachindex(coeffs)
            H3.factors[i] = coeffs[i]
        end
        return H3.factors[7]
    end
end

for t in tspan
    println(Ht(t))
end
```
